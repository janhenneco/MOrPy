This notebook can be used to explore the different `MESA` inlist options without having to start digging in the `MESA` installation directory.
This is especially convenient when setting up a grid in a `.py` script, which is required for using the `mpi` backend as shown in `run_a_grid_with_mpi4py.py`.
For `morpy` projects written in a Jupyter Notebook this can be done in the notebook itself. 

```python
import morpy as mp
import numpy as np
from pathlib import Path
import pandas as pd

MESA_DIR = Path('<PATH_TO_MESA_DIR>')
MESASDK_ROOT = Path('<PATH_TO_MESASDK>')
OMP_THREADS  = 1 # Number of threads used per MESA model. Can be specified on a model-per-model basis when running/launching the models.
MESA_VERSION = 'r24.08.1'

mconf = mp.setup_mesa_env(mesa_dir=MESA_DIR, mesasdk_root=MESASDK_ROOT, omp_num_threads=OMP_THREADS, version_name=MESA_VERSION)
def_opts = mp.build_defaults_options_from_config(mconf)
print('MESA version:', mconf.mesa_version.version_name)
print('     - year:', mconf.mesa_version.year)
print('     - month:', mconf.mesa_version.month)
print('     - patch:', mconf.mesa_version.patch)
print('     - scheme:', mconf.mesa_version.scheme)
print('     - revision:', mconf.mesa_version.revision)

# Pandas tweaks for better looking tables
pd.set_option("display.max_rows", None) # Change this to None if you want to see complete tables. The maximum of 50 is used to keep the notebook output manageable. 
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)
pd.set_option("display.width", None)
```
