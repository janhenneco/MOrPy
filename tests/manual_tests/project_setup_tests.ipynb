```python
from pathlib import Path
import mesa_runner as mr
import pandas as pd
pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)
pd.set_option("display.width", None)
```

```python
mconf = mr.MesaConfig.setup(mesa_dir='/Users/janhenneco/software/mesa/mesa-r24.08.1', version_name='r24.08.1')
def_opts = mr.build_defaults_options_from_config(mconf)

original_workdir_path = '/Users/janhenneco/GitHub/mesa-runner/tests/data_for_tests/r24.08.1/workdirs/high_mass'

og_model = mr.Model.empty(name='original_hm_model')
mr.import_inlists_from_dir(model=og_model, source_dir=original_workdir_path, default_options=def_opts, config=mconf)
```

```python
og_model.inlists['inlist_high_mass'].remove_option('history_interval')
og_model.inlists['inlist_high_mass'].set_option(name='initial_mass',value='299.0',defaults=def_opts)
og_model.inlists['inlist_high_mass'].set_option(name='history_interval',value=1,namelist='controls',defaults=def_opts)
og_model.inlists['inlist_high_mass'].remove_option('overshoot_f(1)')
```

```python
mr.preview(og_model.inlists['inlist_high_mass'].options)
```

```python
new_inlist = mr.Inlist.empty(name='inlist_new',mesa_version=mconf.mesa_version)

controls = {'initial_mass': 15.4,
            'overshoot_f(1)': 0.05,
            'overshoot_f0(1)': 0.005
}

star_job = {'use_suzuki_weak_rates': True,
            'show_net_species_info': '.true.'

}

new_inlist.set_options(defaults=def_opts, controls=controls, star_job=star_job)

new_model = og_model.copy(model_copy_name='new_model')
new_model.add_inlist(new_inlist)
```

```python
print(og_model.created_at)
print(new_model.created_at)

project_dir = mr.create_projectdir(projectdir_path='/Users/janhenneco/GitHub/mesa-runner/tests/data_for_tests/r24.08.1/workdirs/new_proj_dir', mode='star')
new_model_dir = mr.create_modeldir(model=new_model, projectdir_path=project_dir)
```
