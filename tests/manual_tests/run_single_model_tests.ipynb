```python
from pathlib import Path
import mesa_runner as mr
import pandas as pd
pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)
pd.set_option("display.width", None)
```

```python
mconf = mr.setup_mesa_env(mesa_dir='/Users/janhenneco/software/mesa/mesa-r24.08.1', mesasdk_root='/Applications/mesasdk-25.12.1/', omp_num_threads=2, version_name='r24.08.1')
def_opts = mr.build_defaults_options_from_config(mconf)

original_workdir_path = '/Users/janhenneco/GitHub/mesa-runner/tests/data_for_tests/r24.08.1/workdirs/high_mass'

og_model = mr.Model.empty(name='original_hm_model')
mr.import_inlists_from_dir(model=og_model, source_dir=original_workdir_path, default_options=def_opts, config=mconf)

new_model = og_model.copy(model_copy_name='new_model')
new_model.inlists['inlist_high_mass'].set_option(name='max_model_number', value=700, defaults=def_opts, namelist='controls')
new_model.inlists['inlist_high_mass'].set_option(name='initial_mass', value= 20, defaults=def_opts, namelist='controls')
new_model.inlists['inlist_high_mass'].set_option(name='initial_model_number', value= 0, defaults=def_opts, namelist='star_job')
new_model.inlists['inlist_high_mass'].set_option(name='load_model_filename', 
                                                 value= '/Users/janhenneco/GitHub/mesa-runner/tests/data_for_tests/r24.08.1/workdirs/new_proj_dir/zams.mod', 
                                                 defaults=def_opts, 
                                                 namelist='star_job')

project_dir = mr.create_projectdir(projectdir_path='/Users/janhenneco/GitHub/mesa-runner/tests/data_for_tests/r24.08.1/workdirs/new_proj_dir', mode='star')
new_model_dir = mr.create_modeldir(model=new_model, projectdir_path=project_dir, overwrite=True)
```

```python
def_opts.star_job.search('load')
```

```python
out = mr.compile_project(projectdir_path=project_dir, mode="star")
print(out.pretty(2000))
```

```python
result = mr.launch_model_process(model=new_model, default_options=def_opts)
# use `tail -f /path/to/new_model/log.out` to see live logs
```

```python
res = mr.wait_for_model_process(model=new_model, handle=result[0], process=result[1])
print(res.stdout)
```
